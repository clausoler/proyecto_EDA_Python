```python
def limpiar_columnas_numericas(df, columnas):
    """
    Limpia columnas numéricas que contienen comas decimales y las convierte
    al tipo numérico float.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas a limpiar.

    columnas : list
        Lista de columnas que deben convertirse a formato numérico.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con las columnas convertidas a tipo numérico.
    """

    df = df.copy()

    for columna in columnas:

        # Reemplazar comas por puntos
        df[columna] = (
            df[columna]
            .astype(str)
            .str.replace(",", ".", regex=False)
        )

        # Convertir a numérico
        df[columna] = pd.to_numeric(
            df[columna],
            errors="coerce"
        )

    return df
```

```python
# ==============================
# APLICAR FUNCION DE LIMPIEZA
# ==============================
# Se aplica la función sobre las columnas numéricas
# detectadas anteriormente.

df_bank = limpiar_columnas_numericas(
    df_bank,
    columnas_numericas_texto
)
```

```python
# ==============================
# 5.1 REVISION DE VARIABLES BINARIAS
# ==============================
# Se revisan los valores unicos de las variables binarias
# antes de realizar la conversion.

print(df_bank["default"].value_counts(dropna=False))

print(df_bank["housing"].value_counts(dropna=False))

print(df_bank["loan"].value_counts(dropna=False))
```

```python
# ==============================
# 5.2 DEFINICION DE COLUMNAS BINARIAS
# ==============================
# Se almacenan las variables binarias
# que seran transformadas.

columnas_binarias = [
    "default",
    "housing",
    "loan"
]
```

```python
# ==============================
# 5.3 FUNCION PARA CONVERTIR VARIABLES BINARIAS
# ==============================
# Esta funcion transforma variables binarias
# codificadas como 0 y 1 en categorias
# interpretables "No" y "Yes".

def convertir_binarias(df, columnas):
    """
    Convierte variables binarias numericas
    en categorias de texto.

    Parametros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas binarias.

    columnas : list
        Lista de columnas binarias a convertir.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con las variables binarias convertidas.
    """

    df = df.copy()

    mapa = {
        1.0: "Yes",
        0.0: "No"
    }

    for columna in columnas:

        df[columna] = df[columna].map(mapa)

    return df
```

```python
# ==============================
# 5.4 APLICAR CONVERSION DE VARIABLES BINARIAS
# ==============================
# Se aplica la conversion de variables binarias
# para mejorar la interpretabilidad del dataset.

df_bank = convertir_binarias(
    df_bank,
    columnas_binarias
)
```

```python
# ==============================
# 6.2 DICCIONARIO DE MESES EN ESPAÑOL
# ==============================
# Este diccionario servirá para reemplazar
# los nombres de los meses en español por inglés,
# permitiendo que pandas pueda interpretar las fechas.

meses_es_en = {
    "enero": "January",
    "febrero": "February",
    "marzo": "March",
    "abril": "April",
    "mayo": "May",
    "junio": "June",
    "julio": "July",
    "agosto": "August",
    "septiembre": "September",
    "octubre": "October",
    "noviembre": "November",
    "diciembre": "December"
}
```

```python
# ==============================
# 6.3 FUNCION PARA LIMPIAR FECHAS EN ESPAÑOL
# ==============================
# Esta función reemplaza los meses en español
# por meses en inglés y convierte la columna
# al tipo datetime.

def limpiar_fechas_espanol(df, columna):
    """
    Limpia una columna de fechas que contiene meses en español
    y la convierte al tipo datetime.

    Parámetros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la columna de fechas.

    columna : str
        Nombre de la columna que contiene las fechas.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con la columna convertida a datetime.
    """

    df = df.copy()

    # Convertir a string
    df[columna] = df[columna].astype(str)

    # Reemplazar meses en español por inglés
    for mes_es, mes_en in meses_es_en.items():

        df[columna] = df[columna].str.replace(
            mes_es,
            mes_en,
            regex=False
        )

    # Convertir a datetime
    df[columna] = pd.to_datetime(
        df[columna],
        errors="coerce"
    )

    return df
```

```python
# ==============================
# 6.4 APLICAR FUNCION DE LIMPIEZA DE FECHAS
# ==============================
# Se transforma la columna date al formato datetime.

df_bank = limpiar_fechas_espanol(
    df_bank,
    "date"
)
```

```python
# ==============================
# 7.2 FUNCION PARA IMPUTAR NULOS NUMERICOS
# ==============================
# Esta funcion sustituye los valores nulos de columnas numericas
# por la mediana de cada columna.
# Se usa la mediana porque es menos sensible a valores extremos
# que la media y representa mejor el centro de distribuciones sesgadas.

def imputar_nulos_mediana(df, columnas):
    """
    Imputa valores nulos en columnas numericas usando la mediana.

    Parametros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas a imputar.

    columnas : list
        Lista de columnas numericas en las que se sustituiran los nulos.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con los valores nulos imputados por la mediana.
    """

    df = df.copy()

    for columna in columnas:

        mediana = df[columna].median()

        df[columna] = df[columna].fillna(mediana)

    return df
```

```python
# ==============================
# 7.3 FUNCION PARA IMPUTAR NULOS CATEGORICOS
# ==============================
# Esta funcion sustituye los valores nulos de columnas categoricas
# por la etiqueta "unknown".
# Se usa "unknown" porque en variables categoricas el valor nulo
# puede representar informacion desconocida o no registrada.
# De esta forma evitamos inventar una categoria usando la moda.

def imputar_nulos_unknown(df, columnas):
    """
    Imputa valores nulos en columnas categoricas usando la etiqueta "unknown".

    Parametros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas a imputar.

    columnas : list
        Lista de columnas categoricas en las que se sustituiran los nulos.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con los valores nulos imputados como "unknown".
    """

    df = df.copy()

    for columna in columnas:

        df[columna] = df[columna].fillna("unknown")

    return df
```

```python
# ==============================
# 7.4 FUNCION PARA IMPUTAR NULOS CON MODA
# ==============================
# Esta funcion sustituye los valores nulos por el valor mas frecuente
# de cada columna.
# Se puede usar en variables categoricas binarias cuando tiene sentido
# mantener el comportamiento predominante del dataset.

def imputar_nulos_moda(df, columnas):
    """
    Imputa valores nulos usando la moda de cada columna.

    Parametros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas a imputar.

    columnas : list
        Lista de columnas en las que se sustituiran los valores nulos.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con los valores nulos imputados por la moda.
    """

    df = df.copy()

    for columna in columnas:

        moda = df[columna].mode()[0]

        df[columna] = df[columna].fillna(moda)

    return df
```

```python
# ==============================
# 7.5 IMPUTACION NUMERICA
# ==============================
# Se imputan con la mediana las variables numericas con valores nulos.
# La mediana se utiliza porque es menos sensible a valores extremos
# que la media.

columnas_nulos_numericas = [
    "age",
    "cons.price.idx",
    "euribor3m"
]

df_bank = imputar_nulos_mediana(
    df_bank,
    columnas_nulos_numericas
)
```

```python
# ==============================
# 7.6 IMPUTACION CATEGORICA CON UNKNOWN
# ==============================
# Se imputan con "unknown" las variables categoricas donde el valor
# faltante puede interpretarse como informacion desconocida
# o no registrada.

columnas_nulos_unknown = [
    "job",
    "marital",
    "education",
    "default"
]

df_bank = imputar_nulos_unknown(
    df_bank,
    columnas_nulos_unknown
)
```

```python
# ==============================
# 7.7 IMPUTACION DE VARIABLES BINARIAS CON MODA
# ==============================
# Se imputan con la moda las variables binarias housing y loan.
# Al tener un porcentaje bajo de nulos, se usa el valor mas frecuente
# como representacion del comportamiento predominante.

columnas_nulos_moda = [
    "housing",
    "loan"
]

df_bank = imputar_nulos_moda(
    df_bank,
    columnas_nulos_moda
)
```

```python
# ==============================
# 9.11 CREACION DE CUSTOMER_SENIORITY_DAYS
# ==============================
# Se calcula la antiguedad del cliente en dias
# desde su fecha de alta hasta la fecha
# de la campaña de marketing.

df_final["customer_seniority_days"] = (
    df_final["date"] - df_final["Dt_Customer"]
).dt.days
```

```python
# ==============================
# 8.X CREACION DE CUSTOMER_SENIORITY_YEARS
# ==============================
# Se calcula la antiguedad aproximada del cliente
# en años para facilitar la interpretacion.

df_final["customer_seniority_years"] = (
    df_final["customer_seniority_days"] / 365
).round(1)
```

```python
# ==============================
# 8.X CREACION DE CUSTOMER_YEAR
# ==============================
# Se extrae el año en el que el cliente
# se dio de alta en la empresa.

df_final["customer_year"] = (
    df_final["Dt_Customer"].dt.year
)
```

```python
# ==============================
# 8.X CREACION DE CUSTOMER_MONTH
# ==============================
# Se extrae el mes de alta del cliente.

df_final["customer_month"] = (
    df_final["Dt_Customer"].dt.month_name()
)
```
