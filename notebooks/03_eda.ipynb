```python
# ==============================
# 5.2.4 COUNTPLOTS DE VARIABLES CATEGORICAS
# ==============================
# Se generan graficos de barras para visualizar
# la distribucion de las variables categoricas.
# Se excluyen variables identificadoras y variables
# con demasiadas categorias unicas porque no aportan
# valor visual al analisis.

# Variables categoricas a excluir del analisis grafico
variables_excluidas = [
    "id_",
    "ID"
]

# Filtrar variables categoricas validas para graficos
variables_categoricas_graficos = [
    columna
    for columna in variables_categoricas
    if columna not in variables_excluidas
]

for columna in variables_categoricas_graficos:

    # Saltar variables con demasiadas categorias
    if df[columna].nunique() > 20:

        print(
            f"Se omite {columna} por alta cardinalidad."
        )

        continue

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        x=columna,
        order=df[columna].value_counts().index
    )

    plt.title(f"Distribucion de {columna}")

    plt.xticks(rotation=45)

    plt.show()
```

```python
# ==============================
# 6.1.5 ANALISIS DE CORRELACION CON Y
# ==============================
# Se convierte y a formato binario para analizar
# correlaciones con variables numericas.

df["y_binaria"] = np.where(
    df["y"] == "yes",
    1,
    0
)
```

```python
# ==============================
# 6.1.6 CORRELACION NUMERICA CON Y
# ==============================
# Se calcula la correlacion entre variables numericas
# y la variable objetivo binaria.

correlacion_y = (
    df[
        variables_numericas.tolist() + ["y_binaria"]
    ]
    .corr()["y_binaria"]
    .sort_values(ascending=False)
)

correlacion_y
```

```python
# ==============================
# 6.2 VARIABLES CATEGORICAS VS Y
# ==============================
# Se seleccionan las variables categoricas
# para analizar su relacion con la variable objetivo.

variables_categoricas
```

```python
# ==============================
# 6.2.1 TABLAS DE FRECUENCIA CRUZADA
# ==============================
# Se generan tablas cruzadas entre variables categoricas
# y la variable objetivo.
# Se excluyen variables identificadoras porque no aportan
# valor analitico en el analisis bivariado.

# Variables categoricas a excluir
variables_excluidas = [
    "id_",
    "ID"
]

for columna in variables_categoricas:

    # Excluir variables identificadoras
    if columna in variables_excluidas:

        continue

    print("\n==============================")
    print(f"Tabla cruzada: {columna} vs y")
    print("==============================")

    display(
        pd.crosstab(
            df[columna],
            df["y"]
        )
    )
```

```python
# ==============================
# 6.2.2 PORCENTAJES DE SUSCRIPCION
# ==============================
# Se calculan porcentajes de suscripcion
# para cada categoria.

# Variables categoricas a excluir
variables_excluidas = [
    "id_",
    "ID"
]

for columna in variables_categoricas:

    # Excluir variables identificadoras
    if columna in variables_excluidas:

        continue

    print("\n==============================")
    print(f"Porcentaje de suscripcion: {columna}")
    print("==============================")

    tabla_porcentaje = pd.crosstab(
        df[columna],
        df["y"],
        normalize="index"
    ) * 100

    display(
        tabla_porcentaje.round(2)
    )
```

```python
# ==============================
# 6.2.3 COUNTPLOTS VARIABLES CATEGORICAS VS Y
# ==============================
# Se generan graficos de barras agrupados
# segun la variable objetivo.

# Variables categoricas a excluir
variables_excluidas = [
    "id_",
    "ID"
]

for columna in variables_categoricas:

    # Excluir variables identificadoras
    if columna in variables_excluidas:

        continue

    # Excluir variables con demasiadas categorias
    if df[columna].nunique() > 20:

        print(
            f"Se omite {columna} por alta cardinalidad."
        )

        continue

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        x=columna,
        hue="y",
        order=df[columna].value_counts().index
    )

    plt.title(f"{columna} vs y")

    plt.xticks(rotation=45)

    plt.show()
```

```python
# ==============================
# 6.2.4 ANALISIS DE TASA DE CONVERSION
# ==============================
# Se calcula la tasa de conversion promedio
# para cada variable categorica.

# Variables categoricas a excluir
variables_excluidas = [
    "id_",
    "ID"
]

for columna in variables_categoricas:

    # Excluir variables identificadoras
    if columna in variables_excluidas:

        continue

    print("\n==============================")
    print(f"Tasa de conversion: {columna}")
    print("==============================")

    conversion = (
        df.groupby(columna)["y_binaria"]
        .mean()
        .sort_values(ascending=False)
        * 100
    ).round(2)

    display(conversion)
```

```python
# ==============================
# 7.15 TASA DE CONVERSION SEGUN ANTIGUEDAD
# ==============================
# Se calcula la tasa de conversion promedio
# segun grupos de antiguedad del cliente.

bins = [0, 2, 4, 6, 8]

labels = [
    "0-2 años",
    "3-4 años",
    "5-6 años",
    "7-8 años"
]

df["seniority_group"] = pd.cut(
    df["customer_seniority_years"],
    bins=bins,
    labels=labels
)
```

```python
# ==============================
# 7.16 CONVERSION POR GRUPOS DE ANTIGUEDAD
# ==============================
# Se calcula la tasa de conversion
# segun la antiguedad del cliente.

conversion_seniority = (
    df.groupby("seniority_group")["y_binaria"]
    .mean() * 100
).round(2)

conversion_seniority
```

```python
# ==============================
# 7.17 GRAFICO DE CONVERSION POR ANTIGUEDAD
# ==============================
# Se representa la tasa de conversion
# segun grupos de antiguedad.

sns.barplot(
    x=conversion_seniority.index,
    y=conversion_seniority.values
)

plt.title(
    "Tasa de conversion segun antiguedad"
)

plt.xlabel("Antiguedad del cliente")
plt.ylabel("Conversion (%)")

plt.xticks(rotation=45)

plt.show()
```

```python
# ==============================
# 10. DETECCION DE OUTLIERS
# ==============================
# En este bloque se identifican posibles valores atipicos
# en las variables numericas del dataset.
# El objetivo es detectar observaciones extremas
# que puedan influir en el analisis estadistico.
```

```python
# ==============================
# 10.1 SELECCION DE VARIABLES NUMERICAS
# ==============================
# Se revisan las variables numericas
# que seran analizadas.

variables_numericas
```

```python
# ==============================
# 10.2 FUNCION PARA DETECTAR OUTLIERS
# ==============================
# Esta funcion detecta outliers utilizando
# el metodo del rango intercuartil (IQR).

def detectar_outliers_iqr(df, columna):
    """
    Detecta outliers utilizando el metodo IQR.

    Parametros
    ----------
    df : pandas.DataFrame
        DataFrame que contiene la variable.

    columna : str
        Nombre de la columna numerica a analizar.

    Retorna
    -------
    pandas.DataFrame
        DataFrame con los registros considerados outliers.
    """

    q1 = df[columna].quantile(0.25)

    q3 = df[columna].quantile(0.75)

    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr

    limite_superior = q3 + 1.5 * iqr

    outliers = df[
        (df[columna] < limite_inferior)
        |
        (df[columna] > limite_superior)
    ]

    return outliers
```

```python
# ==============================
# 10.3 NUMERO DE OUTLIERS POR VARIABLE
# ==============================
# Se calcula el numero de outliers
# presentes en cada variable numerica.

resumen_outliers = []

for columna in variables_numericas:

    outliers = detectar_outliers_iqr(
        df,
        columna
    )

    porcentaje = (
        len(outliers) / len(df) * 100
    )

    resumen_outliers.append({
        "variable": columna,
        "outliers": len(outliers),
        "porcentaje": round(porcentaje, 2)
    })

resumen_outliers = pd.DataFrame(
    resumen_outliers
)

resumen_outliers.sort_values(
    by="porcentaje",
    ascending=False
)
```

```python
# ==============================
# 10.4 BOXPLOTS DE VARIABLES NUMERICAS
# ==============================
# Se representan boxplots para visualizar
# posibles valores atipicos.

for columna in variables_numericas:

    plt.figure(figsize=(10, 4))

    sns.boxplot(
        x=df[columna]
    )

    plt.title(f"Boxplot de {columna}")

    plt.show()
```
